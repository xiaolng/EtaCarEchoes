# Read in Data

```python
### Get table of light curve data located on website -- two links per filter (for the two ccds)

### filter g
url_a57_g = 'https://stsci-transients.stsci.edu/eta/etalc/results/rod_test/goodpos_g/ec0915/57/ec0915a57_goodpos_g_tmpl523730.html'
url_a54_g = 'https://stsci-transients.stsci.edu/eta/etalc/results/rod_test/goodpos_g/ec0915/54/ec0915a54_goodpos_g_tmpl523730.html'
table_a57_g = get_LCtable(url_a57_g)
table_a54_g = get_LCtable(url_a54_g)
```

```python
### filter r
url_a57_r = 'https://stsci-transients.stsci.edu/eta/etalc/results/rod_test/goodpos_r/ec0915/57/ec0915a57_goodpos_r_tmpl523436.html'
url_a54_r = 'https://stsci-transients.stsci.edu/eta/etalc/results/rod_test/goodpos_r/ec0915/54/ec0915a54_goodpos_r_tmpl523436.html'
table_a57_r = get_LCtable(url_a57_r)
table_a54_r = get_LCtable(url_a54_r)
```

```python
### filter i
url_a57_i = 'https://stsci-transients.stsci.edu/eta/etalc/results/rod_test/goodpos_i/ec0915/57/ec0915a57_goodpos_i_tmpl523435.html'
url_a54_i = 'https://stsci-transients.stsci.edu/eta/etalc/results/rod_test/goodpos_i/ec0915/54/ec0915a54_goodpos_i_tmpl523435.html'
table_a57_i = get_LCtable(url_a57_i)
table_a54_i = get_LCtable(url_a54_i)
```

```python
### filter z
url_a57_z = 'https://stsci-transients.stsci.edu/eta/etalc/results/rod_test/goodpos_z/ec0915/57/ec0915a57_goodpos_z_tmpl523733.html'
url_a54_z = 'https://stsci-transients.stsci.edu/eta/etalc/results/rod_test/goodpos_z/ec0915/54/ec0915a54_goodpos_z_tmpl523733.html'
table_a57_z = get_LCtable(url_a57_z)
table_a54_z = get_LCtable(url_a54_z)
```

```python
### Get the light curve data
### Data divided into groups based on positons/phase
    ## a57 groups: 0 - main, 1 - secondary
    ## a54 groups: 0 - comet, 1 - main peak, 2 - broad LCs, 3 - Armins, 4 - Armins test
### Each segment covers a different phase of the overall light curve


## segment 1
lc_list1_g = get_LCdata(table_a57_g, group=0, indices=[0,1,2,3,4,5])      # some images needed to be excluded
lc_list1_r = get_LCdata(table_a57_r, group=0, indices=[0,1,2,3,4,5])
lc_list1_i = get_LCdata(table_a57_i, group=0, indices=[0,1,2,3,4,5])
lc_list1_z = get_LCdata(table_a57_z, group=0, indices=[0,1,2,3,4,5])

## segment 2
lc_list2_g = get_LCdata(table_a54_g, group=1)
lc_list2_r = get_LCdata(table_a54_r, group=1)
lc_list2_i = get_LCdata(table_a54_i, group=1)
lc_list2_z = get_LCdata(table_a54_z, group=1)

## segment 3
lc_list3_g = get_LCdata(table_a54_g, group=3)
lc_list3_r = get_LCdata(table_a54_r, group=3)
lc_list3_i = get_LCdata(table_a54_i, group=3)
lc_list3_z = get_LCdata(table_a54_z, group=3)

## segment 4
lc_list4_g = get_LCdata(table_a54_g, group=0)
lc_list4_r = get_LCdata(table_a54_r, group=0)
lc_list4_i = get_LCdata(table_a54_i, group=0)
lc_list4_z = get_LCdata(table_a54_z, group=0)
```

```python
### Save tables as csv files -- for rest of analysis

save_dir = 'LC_tables/data'

## reference tables
table_a57_g.to_csv(f'{save_dir}/LCref_a57_g.csv', index=False)
table_a57_r.to_csv(f'{save_dir}/LCref_a57_r.csv', index=False)
table_a57_i.to_csv(f'{save_dir}/LCref_a57_i.csv', index=False)
table_a57_z.to_csv(f'{save_dir}/LCref_a57_z.csv', index=False)

table_a54_g.to_csv(f'{save_dir}/LCref_a54_g.csv', index=False)
table_a54_r.to_csv(f'{save_dir}/LCref_a54_r.csv', index=False)
table_a54_i.to_csv(f'{save_dir}/LCref_a54_i.csv', index=False)
table_a54_z.to_csv(f'{save_dir}/LCref_a54_z.csv', index=False)

## lc data
lc_list1_g.to_csv(f'{save_dir}/LCtable1_g.csv', index=False)
lc_list1_r.to_csv(f'{save_dir}/LCtable1_r.csv', index=False)
lc_list1_i.to_csv(f'{save_dir}/LCtable1_i.csv', index=False)
lc_list1_z.to_csv(f'{save_dir}/LCtable1_z.csv', index=False)

lc_list2_g.to_csv(f'{save_dir}/LCtable2_g.csv', index=False)
lc_list2_r.to_csv(f'{save_dir}/LCtable2_r.csv', index=False)
lc_list2_i.to_csv(f'{save_dir}/LCtable2_i.csv', index=False)
lc_list2_z.to_csv(f'{save_dir}/LCtable2_z.csv', index=False)

lc_list3_g.to_csv(f'{save_dir}/LCtable3_g.csv', index=False)
lc_list3_r.to_csv(f'{save_dir}/LCtable3_r.csv', index=False)
lc_list3_i.to_csv(f'{save_dir}/LCtable3_i.csv', index=False)
lc_list3_z.to_csv(f'{save_dir}/LCtable3_z.csv', index=False)

lc_list4_g.to_csv(f'{save_dir}/LCtable4_g.csv', index=False)
lc_list4_r.to_csv(f'{save_dir}/LCtable4_r.csv', index=False)
lc_list4_i.to_csv(f'{save_dir}/LCtable4_i.csv', index=False)
lc_list4_z.to_csv(f'{save_dir}/LCtable4_z.csv', index=False)
```

# Exclude Bad Images

```python
### Difference images that failed or have huge artifacts

badims_a54 = glob.glob('../../data/testdata/ec0915_tmpl/filt_*/54/bad_ims/*.fits')
badims_a54 = [im.rpartition('/')[-1] for im in badims_a54]
bad_expnums_a54 = [int(re.search('\.(\d+)_ooi_',os.path.basename(im)).groups()[0]) for im in badims_a54]

badims_a57 = glob.glob('../../data/testdata/ec0915_tmpl/filt_*/57/bad_ims/*.fits')
badims_a57 = [im.rpartition('/')[-1] for im in badims_a57]
bad_expnums_a57 = [int(re.search('\.(\d+)_ooi_',os.path.basename(im)).groups()[0]) for im in badims_a57]

pd.DataFrame({'a54': bad_expnums_a54}).to_csv('txt_files/badims_a54.csv', index=False)
pd.DataFrame({'a57': bad_expnums_a57}).to_csv('txt_files/badims_a57.csv', index=False)
```

```python
### Difference images with bad bands

badbs_a54 = glob.glob('../../data/testdata/ec0915_tmpl/filt_*/54/bad_bands/*.fits')
badbs_a54 = [im.rpartition('/')[-1] for im in badbs_a54]
badbs_expnums_a54 = [int(re.search('\.(\d+)_ooi_',os.path.basename(im)).groups()[0]) for im in badbs_a54]
pd.DataFrame({'a54': badbs_expnums_a54}).to_csv('txt_files/badbs_a54.csv', index=False)
```
